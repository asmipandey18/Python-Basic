# OOP

 It means, python allows you to strucuture your code using classes and objects for better organization and reusability

## Class Properties

- Properties are variables that belong to a class.
- It stores the data for each object created from the class

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print("Hello This is " + self.name)

p1 = Person("Asmi", 21)
print(p1.name) # dot operator is used to access the properties of an object
print(p1.age)
````

````python
#@ Can we modify properties

p1.age = 21
print(p1.age)
````

````python
#@ Delete the age property
del p1.age
````

## Python Class Methods
- Methods are functions that belongs to a class.
- It defines the behaviour of object created from the class.

````python
class Person:
    def __init__(self, name):
      self.name = name

    def greet(self):
        print("Hello sir, I am " + self.name)

p1 = Person("Asmi")

p1.greet()
````

````python
## Task 3
## Create a class named Caculator and add following methods
## method add return a + b
## multiple return a * b
## divide return a / b
## subtraction return a - b

class Calculator:

    def add(self, a, b):
        return a + b

    def multiply(self, a, b):
        return a * b

    def divide(self, a, b):
        return a / b

    def subtract(self, a, b):
        return a - b


#object
cal = Calculator()

print(cal.add(10, 5))
print(cal.multiply(10, 5))
print(cal.divide(10, 5))
print(cal.subtract(10, 5))
````

### Multiple Methods

 - A class can have multiple methods that work together

````python
class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = [] # songs = ['song1', 'song2', 'song3']

    def add_song(self, song):
        self.songs.append(song) # p1.songs.append(song 3)
        print(f"Added a song: {song}")

    def remove_song(self, song):
        if song in self.songs:
            self.songs.remove(song)
            print(f"Removed a song: {song}")

    def show_songs(self):
        print(f"Playlist: {self.name}")
        for song in self.songs:
            print(song)

my_playlist = Playlist("Favourites")
my_playlist.add_song("Song 1")
my_playlist.add_song("Song 2")
my_playlist.add_song("Song 3")
my_playlist.show_songs()
````
