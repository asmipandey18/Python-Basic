# Control Flow in Python

## Conditional Statement in Python (if, elif, else)
- if statement
- else statement
- elif statement
- nested
- Few Examples

```python
## if statement
## if <condition>:
     # statement1
     # statement2
age = 20
if age>= 18:
    print("Asmi don")
```

```python
## else
## it executes a block of code if the condition in the 'if statement is False'
age = 16
if age >= 18:
    print("Asmi")
else:
    print("Asmi do ho")
```

```python
## elif It is used in Conditional Statement

# You can use one or more if, elif or else staatement inside another if, elif, or else satatement to create nested condition

## Example: number to be even, odd, negative

num = int(input("Enter a number"))
if num > 0:
    print("The num is positive")
    if num % 2 == 0:
        print("The number is even")
    else:
        print("The number is odd")
elif num == 0:
    print("The number is zero")
else:
    print("The number is negative")
```

```python
# Task1:
## Detrmine if a yera is a leap year or not using a nested condion statement

year = int(input("Enter a year: "))
if year%4 == 0:
  if year%100 == 0:
    if year%400 == 0:
      print("Leap year")
    else:
          print("Not a Leap Year")
  else:
        print("Leap Year")
else:
    print("Not a Leap Year")

```

# Loops in python

## Loops
- For loop
- while loop
- Nested loop
- Examples

```python
## Range is a function (start, stop)
## start inclusive and stop exclusive
## range(start, stop, step)
## step means, how will you move your sequence, step = 2, the sequence will be updated by 2
## range ma, default value chai, 0 bata start hune raicha
range(5)
```

```python
## for loop
# i = 0,1,2,3,4
# 5 will not be counted as range(5) omits exlusive of 5

for i in range(5):
    print(i)
```

```python
for i in range(1, 10):
    print(i)
```

```python
for i in range(1, 10, 2):
    print(i)
```

```python
for i in range(10, 1, -2):
  print(i)
```

```python
## String
str = "Asmi"

for i in str:
  print(i)
```

```python
## while loop
## the while loop continues to execute as long as the condition in True.

count = 0

while count<5:
    print(count)
    count = count +1
```

```python
## Loop Control Statement
## break
## break statement exits the loop prematurely

for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue
## The continmue statement skips the current iteration and continues with the next
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
## pass
## The pass statement is a null operation; it does nothing.

for i in range(5):
    if i==4:
        pass
    print(i)
```

```python
## Nested Loop
## a loop inside a loop

for i in range(3):
    for j in range(2):
      # print("i:", i)
      # print("j:", j)
      print(f"i:{i} and j:{j}")
```

```python
# Task4
## Calculate the sum of first N natural numbers using a while and for loop
## For Loop
n = 10
sum = 0

for i in range(1, n+1):
     sum = sum + i

print(sum)

```

```python
# While loop

n = 10
sum = 0
i = 1

while i <= n:
    sum = sum + i
    i = i + 1
print(sum)
```

```python
# Task5
## Prime number between 1 and 100 using for loop
for num in range(2, 101):

    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num)
```
