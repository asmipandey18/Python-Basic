````python
## Interview questions

# Create a list or from the list below, sort the list without using sort() function
# a = [2, 3, 5, 1, 7]

##I am using the Bubble Sort algorithm. It compares two adjacent elements and swaps them if they are in the wrong orde

##Ascending
a = [2, 3, 5, 1, 7]

for i in range(len(a)):
    for j in range(len(a) - 1):
        if a[j] > a[j + 1]:
            a[j], a[j + 1] = a[j + 1], a[j]

print(a)

##In descending
a = [2, 3, 5, 1, 7]

for i in range(len(a)):
    for j in range(len(a) - 1):
        if a[j] < a[j + 1]:
            a[j], a[j + 1] = a[j + 1], a[j]

print(a)
````

**List Comprehension**
- It is nothing but a single line conditional loop statement
- Basic syntax
  ```expression for item in iterable```

  ``` num ** 2 for num in numbers ```

- with conditional logic [expression for item in iterable if condition]

- Nested List Comprehension [ expression for item1 in iterabale 1 for item 2 in iterable 2]

````python
## List comprehension with conditional logic

lst = []
for i in range(10):
    if i%2 == 0:
          lst.append(i)
print(lst)
````

````python
## same case using list comprehension
lst = [i for i in range(10) if i%2 == 0]
lst
````
