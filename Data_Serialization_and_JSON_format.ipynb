# Data Serialization

Whenever you need to store or send the data through an api, or store the data in database or for backend development, this concept of serialization plays a vital role

Serialization = Converting Python data into a format that can be stored or sent

Deserialization = converting the data into Python data

```python
#@ Data Serialization
import json

data = {
    'name': 'Asmi',
    'age': '21'
}
data
```

```python
type(data)
```

 # JSON format

  a lightweight, text-based file format used to store and transport data

```python
json_str = json.dumps(data)
json_str
```

```python
type(json_str)
```

```python
json_str
```

```python
#@ If I want to access the data of json format, i will load the json data into dict

parse_data = json.loads(json_str)
parse_data
```

```python
type(parse_data)
```

```python
parse_data['name']
```
