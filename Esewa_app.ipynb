```python
import random
```

```python
class DigitalWallet:
  def __init__(self, user_name, initial_balance):
    self.owner = user_name
    self.wallet_id = f"NP--{random.randint(10000, 99999)}"
    self._balance = initial_balance # protected attribute
    self.history = [] # list to store all the transcation
    self.max_limit = 50000 # normal wallet limit: Rs. 50000
```

```python
#@ I need to Inherit from Parent class DigitalWallet

class PremiumWallet(DigitalWallet):
    def __init__(self, user_name, initial_balance=0):
        super().__init__(user_name, initial_balance)
        self.max_limit = 500000 # Premium account limit: 5lakhs
        self.cashback_rate = 0.02 # 2% Cashback on every bill payment
```
