```python
#@ Read a whole file
with open('elevatex.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
#@ Read a file line by line
with open('elevatex.txt', 'r') as file:
    content = file.readlines()
    print(content)
```

```python
#@ Writing a file (Overwriting)
with open('elevatex.txt', 'w') as file:
    file.write('Hello World!\n')
    file.write('I am Asmi pandey \n')
    file.write('I am from Elevate-X')
```

```python
#@ Read a whole file
with open('elevatex.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
#@ What if I dont want to remove the existing content, whereas i am concerned about adding a new lines in a file
#@ without overwriting
#@ we use the mode append 'a'
with open('elevatex.txt', 'a') as file:
    file.write('\nWe are from Bharatpur, Chitwan')
```

```python
#@ Read a whole file
with open('elevatex.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
#@ What if I need to copy a text file source to a destination text file
# copying a file to a new file
with open('elevatex.txt', 'r') as source_file:
    content = source_file.read()
with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
#@ Read a whole file
with open('destination.txt', 'r') as file:
    content = file.read()
    print(content)
```

# Exception Handling

- Understanding Exceptions
  - Run time error handling
  - If run time, logical error aaayo vaney pani, it makes sure that, the corrective actions will be taken without stopping the execution of the program

Exception
- Define an exception as an event, that disrupt the normal flow of a program.
- It occurs when an error is encountered during the program execution.

Common exception includes:
  - ZeroDivisionError: Divide by zero
  - FileNotFoundError: File not found
  - ValueError: Invalid value
  - TypeError: Invalid type

  To handle exception, we use
  - try
  - except
  - else
  - finally

```python
#@ Exception try, except block
try:
  # statement
  a = b
except:
  # statement
  print("The variable has not been defined")
```

```python
#@ try, except and else block

try:
    num = int(input("Enter a number"))
    result = 10/num
except ZeroDivisionError as ex:
    print(ex)
    print("Denominator should always be greater than 0")
except NameError as ex:
    print(ex)
    print("<<<<<<<<<<<Main exception is caught here >>>>>>>>>>>>>>")
else:
    print("No exception")
    print(result)
```

```python
#@ try, except, else and finally
try:
    num = int(input("Enter a number"))
    result = 10/num
except ZeroDivisionError as ex:
    print(ex)
    print("Denominator should always be greater than 0")
except NameError as ex:
    print(ex)
    print("<<<<<<<<<<<Main exception is caught here >>>>>>>>>>>>>>")
else:
    print("No exception")
    print(result)
finally:
  print("Execution complete!!!")
```
