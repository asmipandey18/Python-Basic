# Function
A function is used to reuse code. Instead of writing the same code again and again, you write it once inside a function and call it whenever you need it.

````python
## syntax
def function_name():
    # code
    print("Hello")

# Function call
function_name()
````

````python
## why function?
## A function is used to reuse code. Instead of writing the same code again and again, you write it once inside a function and call it whenever you need it.
def even_odd(num):
    if num%2 == 0:
        print("even")
    else:
        print("odd")

even_odd(0)
````

````python
## variable length argument in python
def add(a, b, c ,d):
  return a+b+c+d

sum = add(1, 2, 3)
print(sum)
````

````python
## In the above example, the numbers of argument is fixed, thus while calling the function, if the no.of argument is not matched, then in this case
## you'll get an error

## To avoid this issue

## Variable length arugment
## positional and keywords arguments
def print_numbers(*args):
    for number in args:
      print(number)

print_numbers(1, 2, 3, 4, 5, 6, 'Elevate-X')
````

````python
def print_numbers(*rebanta):
    for number in rebanta:
      print(number)

print_numbers("asmi", "pandey")
````

In the above example, * means to collext positional arguments whereas rebanta is the variable name

````python
## Similarly, lets see the concept of **kwargs

def details(**info):
    print(info)

details(name = "Asmi", org = "Elevate X")
````

## Examples in Python

````python
# # Example 1 : WAP to convert Cel to Far
def convert(c):
    f = (c * 9/5) + 32
    print("Fahrenheit =", f)

celsius = float(input("Enter Celsius: "))

convert(celsius)
````

````python
# Example 2 :Password Strength Checker
# # Condition for strong password:
#   1. length should be 8 or >8
#   2. pw should contain at least one digit
#   3. atleast one letter should be Uppercase and one should be lowercase
#   4. should use special character like !@#$%^

#   strong
#   else
#   weak password

def check_password(password):

  length = len(password) >= 8
  digit = any(char.isdigit() for char in password)
  upper = any(char.isupper() for char in password)
  lower = any(char.islower() for char in password)
  special = any(char in "@$#&!" for char in password)

  if length and digit and upper and lower and special:
        print("Strong Password")
  else:
        print("Weak Password")

password = input("Enter Password: ")
check_password(password)
````

````python
# Example 3
## Suppose i need to call a function to print.

## function call
## example (100, 200, 300, "elevateX", name = "Rebanta", age = "29", country = "Nepal")

# how does my function defination look like?

# i have position and keyword needs to be printed
# *args: Allows a function to accept any number of positional arguments.
# **kwargs: Allows a function to accept any number of keyword arguments.

def my_function(*args, **kwargs):

    print("Positional arguments:")
    for i in args:
        print(i)

    print("Keyword arguments:")
    for key, value in kwargs.items():
        print(key, "=", value)

my_function(100, 200, 300, "elevateX", name="Asmi", age="21", country="Nepal")
````

````python
# Example 4 : calcualtate factorial of a number
def factorial(num):
    fact = 1

    for i in range(1, num + 1):
        fact = fact * i

    print("Factorial =", fact)

num = int(input("Enter a number: "))
factorial(num)
````

````python
# Example 5 : check whether the string is palindrome or not
text = input("Enter a string: ")

def palindrome(text):
    if text == text[::-1]:
        print("Palindrome")
    else:
        print("Not Palindrome")

palindrome(text)
````

````python
# Example 6: write a function to validate an email address
email = input("Enter Email: ")

def check_email(email):
    if "@" in email and email.endswith(".com"):
        print("Valid Email")
    else:
        print("Invalid Email")

check_email(email)
````

# Lamda Function

Lamda Function in Python:

Lamda functions are small anonymous function defined using the keyword ```lamda``` . It can have any number of arguments but only one expression.
They are commonly used for short operations or as arguements to nigher order function

````python
from ast import Expression
## Syntax

lambda arguments: Expression
````

````python
# example
add = lambda a, b: a+b
print(add(2, 3))
````

````python
#check wheather the number is even or odd using lamda function
num = int(input("Enter a number: "))

check = lambda x: "Even" if x % 2 == 0 else "Odd"
print(check(num))
````

````python
## use map() to convert a list of strings to integers
str_num = ["1", "2", "3"]

# Convert strings to integers
int_num = list(map(int, str_num))

print(int_num)
````

````python
# convert each word to Upper Case

words = ['apple', 'ball', 'cat']

# Convert to uppercase
upper_words = list(map(str.upper, words))

# Print result
print(upper_words)

````
