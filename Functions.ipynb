# Function
A function is used to reuse code. Instead of writing the same code again and again, you write it once inside a function and call it whenever you need it.

```python
## syntax
def function_name():
    # code
    print("Hello")

# Function call
function_name()
```

```python
## why function?
## A function is used to reuse code. Instead of writing the same code again and again, you write it once inside a function and call it whenever you need it.
def even_odd(num):
    if num%2 == 0:
        print("even")
    else:
        print("odd")

even_odd(0)
```

```python
## function with multiple params

def add(a, b):
    return a+b
result = add(2, 3)
print(result)
```

```python
def greet(name):
    print(f"Hello {name} to Elevate-X")

greet("Asmi")
```

```python
## variable length argument in python
def add(a, b, c ,d):
  return a+b+c+d

sum = add(1, 2, 3)
print(sum)
```

```python
## In the above example, the numbers of argument is fixed, thus while calling the function, if the no.of argument is not matched, then in this case
## you'll get an error

## To avoid this issue

## Variable length arugment
## positional and keywords arguments
def print_numbers(*args):
    for number in args:
      print(number)

print_numbers(1, 2, 3, 4, 5, 6, 'Elevate-X')
```

```python
def print_numbers(*rebanta):
    for number in rebanta:
      print(number)

print_numbers("asmi", "pandey")
```

In the above example, * means to collext positional arguments whereas rebanta is the variable name

```python
## Similarly, lets see the concept of **kwargs

def details(**info):
    print(info)

details(name = "Asmi", org = "Elevate X")
```

```python
## when to use *args and **kwargs

# easy way to remember
# order ("momo", "pizza", "coke") // **args

# if i need extra details as well

# order("pizza", size = "large", cheese = True, cold_drink = "Coke") // pizza *args, **kwargs
```

```python
# # Example 1 : WAP to convert Cel to Far
def convert(c):
    f = (c * 9/5) + 32
    print("Fahrenheit =", f)

celsius = float(input("Enter Celsius: "))

convert(celsius)
```

```python
# Example 2 :Password Strength Checker
# # Condition for strong password:
#   1. length should be 8 or >8
#   2. pw should contain at least one digit
#   3. atleast one letter should be Uppercase and one should be lowercase
#   4. should use special character like !@#$%^

#   strong
#   else
#   weak password

def check_password(password):

  length = len(password) >= 8
  digit = any(char.isdigit() for char in password)
  upper = any(char.isupper() for char in password)
  lower = any(char.islower() for char in password)
  special = any(char in "@$#&!" for char in password)

  if length and digit and upper and lower and special:
        print("Strong Password")
  else:
        print("Weak Password")

password = input("Enter Password: ")
check_password(password)
```
