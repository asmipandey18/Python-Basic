# Task 1 using list.

##Create a Python To-Do List with three tasks. Add two new tasks, remove one completed task, check if "Pay bills" exists, and print the remaining tasks.




```python
# To-Do List with three tasks
tasks = ["Pay bills", "Wake up", "Do homework"]

print("Real task:",tasks)

# Add two new tasks
tasks.append("Study Python")
tasks.append("Complete assignment")

print("After Adding Tasks:", tasks)

# Remove one task
tasks.remove("Do homework")

print("After Removing a Task:", tasks)

# Check if "Pay bills" exists
if "Pay bills" in tasks:
    print("Pay bills is in the list.")
else:
    print("Pay bills is not in the list.")

# Print the remaining tasks
print("Remaining Tasks:", tasks)

```

# Task 2

Organizing Student Grades Create a list to store and calculate average grades for students.

```python
# Create a list of student grades
grades = [60, 75, 55, 85, 70]

# Calculate total grades
total = sum(grades)

# Count the number of grades
number_of_grades = len(grades)

# Calculate average
average = total / number_of_grades

# Display the result
print("Student Grades:", grades)
print("Average Grade:", average)
```

# Task 3
Use a Python list to manage a store's inventory by adding, removing, and displaying items.

```python
## answer
# Create a store inventory
inventory = ["Rice", "Sugar", "Oil"]

print("Inventory:", inventory)

# Add a new item
inventory.append("Milk")

print("Adding Item:", inventory)

# Remove an item
inventory.remove("Sugar")

print("Removing Item:", inventory)

# Display the remaining items
print("Items:", inventory)
```

```python
## Question 5 Answer
numbers = [10, 21, 4, 45, 66, 93, 11]

even = 0
odd = 0

for num in numbers:
    if num % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```
